**Lưu lại checkpoint khi train**

```python
from pathlib import Path
import zipfile


CHECKPOINT_DIR = Path(
    "/kaggle/working/PaddleOCR/Model/vi_PP-OCRv6_medium_rec"
)

OUTPUT_ZIP = Path(
    "/kaggle/working/vi_PP-OCRv6_medium_rec_checkpoint.zip"
)

LATEST_FILES = (
    "latest.pdparams",
    "latest.pdopt",
    "latest.states",
)


if not CHECKPOINT_DIR.is_dir():
    raise FileNotFoundError(
        f"Không tìm thấy thư mục checkpoint: {CHECKPOINT_DIR}"
    )


missing_files = [
    file_name
    for file_name in LATEST_FILES
    if not (CHECKPOINT_DIR / file_name).is_file()
]

if missing_files:
    raise FileNotFoundError(
        "Thiếu file checkpoint latest:\n"
        + "\n".join(
            f"- {file_name}"
            for file_name in missing_files
        )
    )


OUTPUT_ZIP.unlink(missing_ok=True)

with zipfile.ZipFile(
    OUTPUT_ZIP,
    mode="w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=6,
) as zip_file:
    for file_name in LATEST_FILES:
        source_file = CHECKPOINT_DIR / file_name
        archive_path = (
            CHECKPOINT_DIR.name
            + "/"
            + file_name
        )

        zip_file.write(
            source_file,
            arcname=archive_path,
        )

        print("Added:", file_name)


with zipfile.ZipFile(OUTPUT_ZIP, mode="r") as zip_file:
    bad_file = zip_file.testzip()

    if bad_file is not None:
        raise RuntimeError(
            f"File ZIP bị lỗi tại: {bad_file}"
        )


print("\nCreated:", OUTPUT_ZIP)
print(
    "Size:",
    f"{OUTPUT_ZIP.stat().st_size / (1024 ** 2):.2f} MB",
)
```

# **9) Clean Output**

```python
from pathlib import Path
import shutil


WORKING_DIR = Path("/kaggle/working")

KEEP_FILES = {
    WORKING_DIR / "PP-OCRv6_medium_rec_infer.zip",
    WORKING_DIR / "vi_PP-OCRv6_medium_rec_checkpoint.zip",
}


# Kiểm tra đầy đủ các file cần giữ trước khi xóa
missing_files = [
    path
    for path in KEEP_FILES
    if not path.is_file()
]

if missing_files:
    raise FileNotFoundError(
        "Không tìm thấy các file cần giữ:\n"
        + "\n".join(
            str(path)
            for path in missing_files
        )
    )


keep_resolved = {
    path.resolve()
    for path in KEEP_FILES
}


for path in WORKING_DIR.iterdir():
    # Giữ lại hai file ZIP
    if path.resolve() in keep_resolved:
        continue

    try:
        if path.is_symlink() or path.is_file():
            path.unlink()
            print("Deleted file:", path)

        elif path.is_dir():
            shutil.rmtree(path)
            print("Deleted folder:", path)

    except Exception as exc:
        print(f"Delete failed: {path} | {exc}")


print("\nCleanup completed.")

print("\nKept files:")
for path in sorted(KEEP_FILES):
    print(
        "-",
        path,
        f"({path.stat().st_size / (1024 ** 2):.2f} MB)",
    )

print("\nRemaining items:")
for path in sorted(WORKING_DIR.iterdir()):
    print("-", path)
```
